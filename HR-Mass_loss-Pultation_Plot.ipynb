```python
import os
import pandas as pd
import matplotlib.pyplot as plt
from matplotlib.gridspec import GridSpec

history_path = os.path.expanduser('~/file_path/history.data')

data_start_row = 5

try:
    history_data = pd.read_csv(history_path, skiprows=data_start_row, sep=r'\s+', engine='python')
except FileNotFoundError:
    print(f"ERROR: File tidak ditemukan di '{history_path}'. Periksa kembali path Anda.")
    exit()

try:
    Teff = history_data['log_Teff']
    L = history_data['log_L']
    time_myr = history_data['star_age'] / 1e6 
    pulsation_period = history_data['logP']
    wind_mass_loss = history_data['log_abs_mdot']
except KeyError as e:
    print(f"ERROR: Kolom {e} tidak ditemukan. Kolom yang tersedia: {history_data.columns.tolist()}")
    exit()
    
fig = plt.figure(figsize=(15, 8))
gs = GridSpec(2, 2, figure=fig)

# A Panel: HR Diagram
ax_hr = fig.add_subplot(gs[:, 0])
ax_hr.plot(Teff, L, color='black', label='1.85 M$_\odot$ (0.016/0.032)')
ax_hr.set_xlabel(r'log (T$_\mathrm{eff}$/K)', fontsize=14)
ax_hr.set_ylabel(r'log (L/L$_\odot$)', fontsize=14)
ax_hr.set_title('HR Diagram', fontsize=16)
ax_hr.invert_xaxis()
ax_hr.grid(True, linestyle='--', alpha=0.6)
ax_hr.legend()
ax_hr.text(0.05, 0.95, '(a)', transform=ax_hr.transAxes, fontsize=16, 
           fontweight='bold', va='bottom', ha='left', color='black')

# B panel: Pultation Overtime
ax_pulsation = fig.add_subplot(gs[0, 1])
ax_pulsation.plot(time_myr, pulsation_period, color='blue', label='Pulsation Period')
ax_pulsation.set_ylabel(r'log (P/days)', fontsize=14)
ax_pulsation.set_title('Pulsation Period Over Time', fontsize=16)
ax_pulsation.grid(True, linestyle='--', alpha=0.6) # Koreksi dari kode Anda
ax_pulsation.legend()
# Menambahkan label panel (b)
ax_pulsation.text(0.05, 0.95, '(b)', transform=ax_pulsation.transAxes, fontsize=16, 
                  fontweight='bold', va='top', ha='left', color='blue')


# C panel: Wind Mass Loss Overtime
ax_wind_loss = fig.add_subplot(gs[1, 1], sharex=ax_pulsation) # Menghubungkan sumbu x dengan plot di atasnya
ax_wind_loss.plot(time_myr, wind_mass_loss, color='orange', label='Wind Mass-Loss')
ax_wind_loss.set_xlabel('Time (Myr)', fontsize=14)
ax_wind_loss.set_ylabel(r'log($|\dot{M}|$/M$_\odot$ yr$^{-1}$)', fontsize=14)
ax_wind_loss.set_title('Wind Mass-Loss Over Time', fontsize=16)
ax_wind_loss.grid(True, linestyle='--', alpha=0.6)
ax_wind_loss.legend()
ax_wind_loss.text(0.05, 0.95, '(c)', transform=ax_wind_loss.transAxes, fontsize=16, 
                  fontweight='bold', va='top', ha='left', color='orange')

plt.tight_layout()
output_path = os.path.expanduser('~/file_path/Plot_Panel_Msun_Overshoot.png')
plt.savefig(output_path, dpi=1200)
print(f"Plot dengan panel telah disimpan di: {output_path}")

plt.show()
```
